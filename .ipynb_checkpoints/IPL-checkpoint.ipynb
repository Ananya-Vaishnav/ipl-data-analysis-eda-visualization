```python
#Loading required libraries
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
#Loading the ipl matches dataset
ipl = pd.read_csv('matches.csv')
```

```python
#glance of first 5 records
ipl.head()

#win_by_runs is when you first batted and won the match
#win_by_wickets is when you first fielded, then batted, and won the match
```

```python
#number of rows and columns in dataset
ipl.shape
```

```python
#freq of most man of the match awards
ipl['player_of_match'].value_counts()
```

```python
#top 10 players with the most man of the match awards
ipl['player_of_match'].value_counts()[0:10]
```

```python
#top 5 players with most man of the match awards
ipl['player_of_match'].value_counts()[0:5]
```

```python
#showing just names of top 5 man of the match players
list(ipl['player_of_match'].value_counts()[0:5].keys())
```

```python
#plotting top 5 man of the match players in bar-plot
plt.figure(figsize = (8,5))
plt.bar(list(ipl['player_of_match'].value_counts()[0:5].keys()), list(ipl['player_of_match'].value_counts()[0:5]), color ='g')
plt.show()
```

```python
ipl['result'].value_counts()
```

```python
#finding out the number of toss wins w.r.t each team
ipl['toss_winner'].value_counts()
```

```python
#Extracting the records where a team WON, batting first
batting_first = ipl[ipl['win_by_runs'] != 0]
```

```python
#looking at the head of the prev cell
batting_first.head()
```

```python
#Making a Histogram
plt.figure(figsize=(5,5))
plt.hist(batting_first['win_by_runs'])
plt.title('Distribution of Runs')
plt.xlabel('Runs')
plt.show()
```

```python
# Finding out the number of wins w.r.t each team after batting first
batting_first['winner'].value_counts()
```

```python
# Making a bar-plot for top 3 teams with most wins after batting first
plt.figure(figsize=(6,6))
plt.bar(list(batting_first['winner'].value_counts()[0:3].keys()),list(batting_first['winner'].value_counts()[0:3]), color = ['blue','yellow','red'])
plt.show()
```

```python
plt.figure(figsize=(20,20))
plt.pie(
    list(batting_first['winner'].value_counts()),
    labels=list(batting_first['winner'].value_counts().keys()),
    autopct = '%0.1f%%',
    textprops = {'fontsize' : 15})        #textprops is used to increase the size of %ages and labels
plt.show()
```

```python
#Extracting the records where a team WON, batting second
batting_second = ipl[ipl['win_by_wickets'] != 0]
```

```python
#lloking at the head of prev cell
batting_second.head()
```

```python
#Making a histogram for frequency of wins w.r.t  number of wickets
plt.figure(figsize=(5,5))
plt.hist(batting_second['win_by_wickets'],bins=30)
plt.show()
```

```python
#Finding out the frequency of number of wins w.r.t each team after batting second
batting_second['winner'].value_counts()
```

```python
# Making a bar-plot for top 3 teams with most wins after batting second
plt.figure(figsize=(6,5))
plt.bar(list(batting_first['winner'].value_counts()[0:3].keys()), list(batting_first['winner'].value_counts()[0:3]), color = ['purple', 'blue', 'red'])
plt.show()
```

```python
# Making a pie-chart for distribution of most wins after batting second
plt.figure(figsize=(10,10))
plt.pie(
    list(batting_second['winner'].value_counts()),
    labels = list(batting_second['winner'].value_counts().keys()),
    autopct = '%0.1f%%')
plt.show()
```
